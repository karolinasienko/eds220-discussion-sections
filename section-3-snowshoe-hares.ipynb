a. Who assembled this data, and why?
> The Bonanza Creek Long-Term Ecological Research (LTER) program to study Snowshoe hares, Lepus americanus, and thier population fluctuations of 8-11 years.

b. What does a single row in the data represent?
> One single row represents one hare.

c. When and where were the observations collected?
> The observations were collected at 5 locales in the Tanana Valley in Alaska, from Tok in the east to Clear in the west from 1999 to 2002.

d. Does the dataset contain sensitive data?
> No, this dataset does not contain sensitive data.

e. What study protocols were used to collect the data?
> There are no explicitly listed protocols describing how the data was collected.

f. What are the units for the hares’ weight?
> The hares' weights are recorded in grams.

g. Does the metadata say how missing data is encoded?
> The metadata does not say how missing data is encoded, they do say in the disclaimer: "While substantial efforts are made to ensure the accuracy of data and documentation contained in this Data Set, complete accuracy of data and metadata cannot be guaranteed."
