## 6. Clean values

a. Create a new column called `sex_simple` where:
* `F`, `f`, and `f_` (f with a space) in the `sex` column get assigned to ‘female’,
* `M`, `m`, and `m_` (m with a space) get assigned to ‘male’, and
missing values stay `NaN`.

You decide what to do with the other codes (`?`, `F?`, `f?`, `M?`, `m?`, `pf`). In a markdown cell, write your choice and why you made it.

> We decided to make the f? codes `female?` and m? codes `male?`. For just the ? and `pf`, we made those `NaN` values.

```python
hares['sex_simple'] = hares['sex'].replace({
    'F' : 'female',
    'f' : 'female',
    'f ': 'female',
    'M' : 'male',
    'm' : 'male',
    'm ': 'male',
    'F?': 'female?',
    'f?': 'female?',
    'M?': 'male?',
    'm?': 'male?',
    '?' : np.nan,
    'pf': np.nan
}
)
```

b. Check the counts of unique values (including NAs) in the new `sex_simple` column.

```python
hares['sex_simple'].value_counts(dropna = False)
```
